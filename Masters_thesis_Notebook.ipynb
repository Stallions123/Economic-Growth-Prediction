# Applying LSTM For Time Series Forecasting

```python
# Import modules and packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import datetime as dt
from datetime import datetime

from keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint, TensorBoard

%matplotlib inline
```

```python
dataset = dataset.set_index('Year')
```

```python
dataset.drop(['Country_Name'],axis=1,inplace=True)
```

```python
dataset.head()
```

```python
dataset_copy = dataset.copy()
```

```python
def samples(dataset,window_size = 6):
    df_as_np = dataset.to_numpy()
    X = []
    y = []
    for i in range(len(df_as_np)-window_size):
        row = [a for a in df_as_np[i:i+window_size]]
        X.append(row)
        label = df_as_np[i+window_size][1]
        y.append(label)
    return np.array(X),np.array(y)
```

```python
X2,y2 = samples(dataset)
X2.shape,y2.shape
```

```python
#Splitting the data
X2_train,y2_train = X2[:45],y2[:45]
X2_val , y2_val = X2[45:50],y2[45:50]
X2_test , y2_test = X2[50:],y2[50:]
```

```python
#Normalizing the data
temp_training_mean = np.mean(X2_train[:,:,0])
temp_training_std = np.std(X2_train[:,:,0])

def preprocess(X):
    X[:,:,0] = (X[:,:,0] - temp_training_mean)/temp_training_std
    return X   
```

```python
preprocess(X2_train)
preprocess(X2_val)
preprocess(X2_test)
```

```python
#Building the model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import *
from tensorflow.keras.callbacks import ModelCheckpoint
from tensorflow.keras.losses import MeanSquaredError
from tensorflow.keras.metrics import RootMeanSquaredError
from tensorflow.keras.optimizers import Adam
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import LSTM
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing import sequence

model = tf.keras.Sequential()
model.add(tf.keras.layers.LSTM(128,input_shape=(6,10),return_sequences = True))
model.add(tf.keras.layers.LeakyReLU(alpha=0.5))
model.add(tf.keras.layers.LSTM(128,return_sequences=True))
model.add(tf.keras.layers.LeakyReLU(alpha=0.5))
model.add(tf.keras.layers.Dropout(0.3))
model.add(tf.keras.layers.LSTM(64,return_sequences=False))
model.add(tf.keras.layers.Dropout(0.3))
model.add(tf.keras.layers.Dense(1))

model.summary()
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(monitor = 'val_loss',
                                                 patience = 2,
                                                 mode = 'min')

model.compile(loss = tf.losses.MeanSquaredError(),
             optimizer = tf.optimizers.Adam(),
             metrics = [tf.metrics.MeanAbsoluteError()])
```

```python
#Model Compilation
cp1 = ModelCheckpoint('model1/', save_best_only=True)
model.compile(loss=MeanSquaredError(), optimizer=Adam(learning_rate=0.0001), metrics=[RootMeanSquaredError()])
```

```python
#Model Fitting
model.fit(X2_train, y2_train, validation_data=(X2_val, y2_val), epochs=10, callbacks=[cp1])
```

```python
def postprocess_p(arr):
    arr = (arr*temp_training_std) + temp_training_mean
    return arr

def plot_predictions2(model, X, y, start=0, end=100):
    predictions = model.predict(X)
    p_preds = postprocess_p(predictions[:,0])
    p_actuals = postprocess_p(y)
    df = pd.DataFrame(data={
                          'GDP Predictions': p_preds,
                          'GDP Actuals': p_actuals
                          })
    plt.plot(df['GDP Predictions'][start:end])
    plt.plot(df['GDP Actuals'][start:end])
    return df[start:end]
```

```python
post_processed_df = plot_predictions2(model, X2_test, y2_test)
post_processed_df
```

```python
from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(post_processed_df["GDP Actuals"], post_processed_df["GDP Predictions"]))
```

# Multivariate Forecasting Using Darts (Using GDP Per Capita) Exponential Smoothing

```python
import darts
from darts import TimeSeries
```

```python
data_gdp = dataset["GDP_per_capita_current_US$)"]
```

```python
from darts import TimeSeries
from darts.models import ExponentialSmoothing
from darts.metrics import mape,rmse,smape,mae,mse,rmsle
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure
train_ts = data_gdp[:45]
test_ts = data_gdp[45:]
```

```python
from darts.metrics import rmsle
```

```python
type(train_ts)
```

```python
#Converting the series to dataframe
train_ts = train_ts.to_frame()
```

```python
test_ts = test_ts.to_frame()
```

```python
tr_TS = TimeSeries.from_dataframe(train_ts)
ts_TS = TimeSeries.from_dataframe(test_ts)
```

```python
tr_TS
```

```python
#Exponential Smoothing
model = ExponentialSmoothing(seasonal_periods=2)
model.fit(tr_TS)
pred_TS = model.predict(len(ts_TS))

figure(figsize=(12,6),dpi=80)
tr_TS.plot()
ts_TS.plot(label='Test')
pred_TS.plot(label='Prrediction')
mape_score = mape(ts_TS,pred_TS)
rmse_score = rmse(ts_TS,pred_TS)
smape_score = smape(ts_TS,pred_TS)
mse_score = mse(ts_TS,pred_TS)
rmsle_score = rmsle(ts_TS,pred_TS)
print(rmsle_score)
print("mape={:.3f},rmse={:.3f},smape={:.3f}".format(mape_score,rmse_score,smape_score))
```

# Multivariate Forecasting Using Theta Model

```python
from darts.models import Theta

thetas = 2 - np.linspace(-10,10,50)

best_mape = float('inf')
best_theta = 0

for theta in thetas:
    model = Theta(theta)
    model.fit(tr_TS)
    pred_TS = model.predict(len(ts_TS))
    res = mape(ts_TS,pred_TS)
    if res < best_mape:
        best_mape = res
        best_theta = theta
        
######### Best Theta Model ########

best_theta_model = Theta(best_theta)
best_theta_model.fit(tr_TS)
pred_TS = best_theta_model.predict(len(ts_TS))


figure(figsize=(12,6),dpi=80)
tr_TS.plot(label='Train')
ts_TS.plot(label='Test')
pred_TS.plot(label='Prediction')

mape_score = mape(ts_TS,pred_TS)
rmse_score = rmse(ts_TS,pred_TS)
smape_score = smape(ts_TS,pred_TS)
rmsle_score = rmsle(ts_TS,pred_TS)
print("rmsle={:.3f}".format(rmsle_score))
print("mape={:.3f},rmse={:.3f},smape={:.3f}".format(mape_score,rmse_score,smape_score))    
```
